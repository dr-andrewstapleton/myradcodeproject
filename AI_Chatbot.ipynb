```python
# Once the package has been installed we need to import it into our code of use
import google.generativeai as ai
```

```python
# Next we'll need to access the 'ai' object via an interface (API); specifcially an Application Programming Interface
# And to do that we need a key.  Each person will need to have their own key that 'unlocks' access to the Google Gemini Chatbot.

# To obtain a key, head to https://aistudio.google.com/app/apikey.
# Then click 'Create API Key'.
# Ensure that you keep the key secret and don't share it.

# Copy the key and then paste it in the area provided in the code below.
# This will assign the variable 'API_KEY' the string value of your API key.

# API Key
# API_KEY = 'insert your API key here'
# Replace the string text above with your actual API key and uncomment



# Configure the API
ai.configure(api_key=API_KEY)
```

```python
# Create a new model

# This is where we establish the Generative Model for our chatbot
# The list of Gemini models is available at https://ai.google.dev/gemini-api/docs/models

# The model we will choose is Gemini 3.8 Flash
# https://ai.google.dev/gemini-api/docs/models/gemini-3.8-flash


model = ai.GenerativeModel("gemini-3.8-flash")
chat = model.start_chat()
```

```python
# Loop until the user enters the text 'bye'
while True:
    # Get user input
    message = input('You: ')
    # End chat if 'bye' is entered
    if message.lower() == 'bye':
        print('Chatbot: Goodbye!')
        break
    # Send message to AI and print the response, including adding a newline after the response
    response = chat.send_message(message)
    print('Chatbot:', response.text, '\n')


# Once you execute this code cell your chatbot will run and you will be able to converse with it
# To quit the chat enter the text 'bye' as detailed in the code above

# Note you may initially receive an error similar to:
# ERROR:tornado.access:503 POST /v1beta/models/gemini-3.8-flash:generateContent?%24alt=json%3Benum-encoding%3Dint (::1) 5471.49ms
# This is a known issue due to latency when using python notebooks in Google Colab.
# It should only appear once and then your chatbot will function normally
```
